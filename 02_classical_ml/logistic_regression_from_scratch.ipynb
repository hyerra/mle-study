```python
import numpy as np
from sklearn.datasets import load_breast_cancer
from sklearn import linear_model

np.random.seed(15)

X, y_labels = load_breast_cancer(return_X_y=True)
N = X.shape[0]

W = np.random.randn(30) * 0.1
b = np.random.randn(1) * 0.1

def logistic(x):
    return 1 / (1 + np.exp(-x))

def avg_log_likelihood(X: np.ndarray, y: np.ndarray, W: np.ndarray, b: np.ndarray):
    z = W @ X.T + b
    return (1/N) * np.sum(y * (-np.logaddexp(0, -z)) + (1 - y) * (-np.logaddexp(0, z)))

def maximize_avg_log_likelihood(X: np.ndarray, y: np.ndarray, W: np.ndarray, b: np.ndarray, alpha):
    while True:
        prev_l = avg_log_likelihood(X, y, W, b)
        h = logistic(W @ X.T + b)
        W += alpha/N * (y - h) @ X
        b += alpha/N * (y - h) @ np.ones_like(h)
        l = avg_log_likelihood(X, y, W, b)
        if np.abs(prev_l - l) < 1E-6:
            break
    return W, b


lg = linear_model.LogisticRegression().fit(X, y_labels)

maximize_avg_log_likelihood(X, y_labels, W, b, 0.01)

y_pred_sk = lg.predict(X)
y_pred_custom = (logistic(W @ X.T + b) >= 0.5).astype(int)
agreement = np.mean(y_pred_custom == y_pred_sk)
print(f"Agreement: {agreement:.4f}")
```
